## 2. Exploring patterns
I chose a small number of summaries and figures that relate to my research questions, rather than plotting every variable.

### Distributions of key numerical variables

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, col, label in zip(axes, ['Age', 'BMI', 'MMSE'],
                          ['Age (years)', 'BMI (kg/m²)', 'MMSE score (0-30)']):
    ax.hist(df[col], bins=30, edgecolor='white')
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel(label)
    ax.set_ylabel('Number of patients')
plt.tight_layout()
plt.show()
```

All three histograms are close to flat. In real patients, BMI usually clusters around the middle and MMSE is usually skewed towards high scores, since most people are not severely impaired. A flat shape suggests the values were drawn evenly across a range.

### Correlation matrix (numerical variables only)
I only included genuinely numerical variables. Binary and coded variables such as Gender or Ethnicity are left out, because a correlation on a category code is not a meaningful measurement.

```python
corr = df[numerical].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1,
            square=True, annot_kws={'size': 8})
plt.title('Correlation matrix of numerical variables (Pearson)')
plt.tight_layout()
plt.show()
```

Every correlation is between -0.05 and 0.05, so essentially zero. This includes pairs that are strongly linked in real patients, such as systolic and diastolic blood pressure, or total and LDL cholesterol. Even MMSE, Functional Assessment and ADL, which all measure impairment, are not related to each other. This is the strongest sign that the columns were generated independently of each other.

### Test scores by diagnosis

```python
df['Diagnosis_label'] = df['Diagnosis'].map({0: "No Alzheimer's", 1: "Alzheimer's"})

print(df.groupby('Diagnosis')[['MMSE', 'FunctionalAssessment', 'ADL']].median())

tests = {'MMSE': 'MMSE score (0-30)',
         'FunctionalAssessment': 'Functional assessment (0-10)',
         'ADL': 'ADL score (0-10)'}
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, (col, label) in zip(axes, tests.items()):
    sns.boxplot(data=df, x='Diagnosis_label', y=col, ax=ax)
    ax.set_title(col)
    ax.set_xlabel('')
    ax.set_ylabel(label)
fig.suptitle('Cognitive and functional test scores by diagnosis')
plt.tight_layout()
plt.show()
```

Diagnosed patients score clearly lower on all three tests. Median MMSE is 17.2 for undiagnosed and 11.6 for diagnosed patients, Functional Assessment is 6.2 vs 3.3, and ADL is 6.1 vs 3.2. These are the variables that separate the two groups most clearly.

### Lifestyle factors by diagnosis (Research Question 1)

```python
lifestyle = {'BMI': 'BMI (kg/m²)',
             'AlcoholConsumption': 'Alcohol (units/week)',
             'PhysicalActivity': 'Physical activity (hours/week)',
             'DietQuality': 'Diet quality (score 0-10)',
             'SleepQuality': 'Sleep quality (score 4-10)'}

print(df.groupby('Diagnosis')[list(lifestyle)].median())

fig, axes = plt.subplots(1, 5, figsize=(20, 5))
for ax, (col, label) in zip(axes, lifestyle.items()):
    sns.boxplot(data=df, x='Diagnosis_label', y=col, ax=ax)
    ax.set_title(col)
    ax.set_xlabel('')
    ax.set_ylabel(label)
fig.suptitle('Q1: Lifestyle factors by diagnosis')
plt.tight_layout()
plt.show()
```

The boxes for the two groups almost completely overlap. Median BMI is 27.6 vs 28.0, and physical activity, diet quality and alcohol consumption are nearly identical. Sleep quality is slightly lower in diagnosed patients (7.2 vs 6.9), but the difference is small.

### Smoking and medical history by diagnosis (Research Question 2)
These variables are all binary, so I used crosstabs showing the percentage diagnosed in each group, instead of correlation.

```python
history = ['Smoking', 'FamilyHistoryAlzheimers', 'CardiovascularDisease',
           'Diabetes', 'Depression', 'HeadInjury', 'Hypertension']

for col in history:
    print(pd.crosstab(df[col], df['Diagnosis'], normalize='index').round(2), "\n")
```

```python
rows = []
for col in history:
    for value, name in [(0, 'No'), (1, 'Yes')]:
        pct = df.loc[df[col] == value, 'Diagnosis'].mean() * 100
        rows.append({'Condition': col, 'Has condition': name, '% diagnosed': pct})
hist_df = pd.DataFrame(rows)

plt.figure(figsize=(13, 5))
sns.barplot(data=hist_df, x='Condition', y='% diagnosed', hue='Has condition')
plt.title("Q2: Percentage diagnosed with Alzheimer's by smoking and medical history")
plt.xlabel('')
plt.ylabel('% diagnosed')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

The differences are small. Patients with cardiovascular disease or hypertension are diagnosed slightly more often (39% vs 35%). Surprisingly, patients with a family history of Alzheimer's, diabetes or a head injury are diagnosed slightly less often (around 32 to 33% vs 36%), which is the opposite of what the literature would predict. Smoking makes almost no difference (35% vs 36%).

### Symptoms by diagnosis and by test score (Research Question 3)

```python
symptoms = ['MemoryComplaints', 'BehavioralProblems', 'Confusion', 'Disorientation',
            'PersonalityChanges', 'DifficultyCompletingTasks', 'Forgetfulness']

# % diagnosed without [0] and with [1] each symptom
for col in symptoms:
    print(col, pd.crosstab(df[col], df['Diagnosis'], normalize='index')[1].round(2).tolist())
```

```python
# Median test scores without (0) and with (1) each symptom
for col in symptoms:
    print(df.groupby(col)[['MMSE', 'FunctionalAssessment', 'ADL']].median().round(2), "\n")
```

```python
fig, axes = plt.subplots(2, 4, figsize=(20, 9))
axes = axes.flatten()
for ax, col in zip(axes, symptoms):
    sns.boxplot(data=df, x=col, y='MMSE', ax=ax)
    ax.set_title(col)
    ax.set_xticks([0, 1])
    ax.set_xticklabels(['No', 'Yes'])
    ax.set_xlabel('')
    ax.set_ylabel('MMSE score (0-30)')
axes[-1].axis('off')
fig.suptitle('Q3: MMSE score by reported symptom')
plt.tight_layout()
plt.show()
```

Two symptoms are strongly linked to diagnosis. 64% of patients with memory complaints are diagnosed compared with 28% without, and 60% of patients with behavioral problems are diagnosed compared with 31% without. The other five symptoms make almost no difference to diagnosis.

None of the symptoms are clearly linked to lower test scores. MMSE is about the same, or even slightly higher, in patients who report a symptom. This is surprising, because memory complaints are associated with diagnosis and diagnosis is associated with lower MMSE, yet memory complaints are not associated with MMSE. In real data, these three would normally line up.

## 3. Research questions

### Research Question 1: Lifestyle factors
**Research question:** Do patients diagnosed with Alzheimer's differ from undiagnosed patients in lifestyle factors (BMI, physical activity, diet quality, sleep quality, alcohol consumption, smoking)?

**Rationale:** Lifestyle factors are modifiable. If any of them is linked to diagnosis, that could matter for prevention.

**Variables:** Diagnosis is the comparison variable. BMI, PhysicalActivity, DietQuality, SleepQuality and AlcoholConsumption are numerical predictors. Smoking is a binary predictor.

**Analysis plan:** Compare the median and IQR of each numerical lifestyle variable between diagnosis groups. For smoking, use a crosstab showing the proportion diagnosed among smokers and non-smokers.

**Visualization:** Side-by-side boxplots of each lifestyle variable split by diagnosis, and a bar chart of the proportion diagnosed by smoking status.

**Limitations:** Lifestyle is measured once, so we can't tell whether it came before the disease or changed because of it. The units and scales for diet, sleep and activity are not documented. Age may confound the results.

### Research Question 2: Medical history
**Research question:** Is a history of family Alzheimer's, cardiovascular disease, diabetes, depression, head injury or hypertension linked to a higher proportion of Alzheimer's diagnoses?

**Rationale:** Several of these conditions are discussed in the literature as possible risk factors. This checks whether the pattern shows up in this dataset.

**Variables:** Diagnosis is the comparison variable. The six binary history variables are the predictors.

**Analysis plan:** For each history variable, make a crosstab with diagnosis and compare the percentage diagnosed with and without the condition.

**Visualization:** A grouped bar chart showing the percentage diagnosed for "yes" vs "no" on each of the six conditions.

**Limitations:** History is recorded as yes/no only, with no severity or timing. Conditions may overlap within patients, so their effects are hard to separate. The 65/35 diagnosis imbalance should be kept in mind when reading proportions.

### Research Question 3: Symptoms and functional tests
**Research question:** Do patients who report cognitive or behavioral symptoms (memory complaints, behavioral problems, confusion, disorientation, personality changes, difficulty completing tasks, forgetfulness) score lower on functional and cognitive tests (MMSE, Functional Assessment, ADL)?

**Rationale:** If reported symptoms match test performance, symptoms could be a cheap early screening signal. If they don't, that says something about how reliable self-reports are.

**Variables:** MMSE, FunctionalAssessment and ADL are the numerical outcomes. The seven symptom variables are binary groups. Here diagnosis is not the outcome.

**Analysis plan:** For each symptom, compare the mean and median test scores between patients with and without it. Because the symptoms are binary, use group comparisons rather than correlation.

**Visualization:** Boxplots of MMSE (and the other two tests) split by each symptom, laid out in a grid.

**Limitations:** The scoring direction and units of FunctionalAssessment and ADL are not documented, so it's not clear whether higher is better. Symptoms may be self-reported or reported by family, which adds noise. The data shows signs of being synthetic, which could limit how well any patterns reflect real patients.

## Data-quality issues and uncertainties

- **No missing values.** The data is complete and has no duplicates, which is unusual for real clinical records.
- **Constant column.** DoctorInCharge is "XXXConfid" in every row and was dropped.
- **Unusual precision.** Continuous variables have as many unique values as there are patients, with many decimal places. Real measurements like BMI or blood pressure are usually rounded.
- **Flat distributions.** Histograms of BMI, age and MMSE are almost flat, and the mean and median of every variable are nearly equal. This looks like values drawn evenly across a set range.
- **No correlations between related measures.** All correlations are close to zero, even between systolic and diastolic blood pressure, and between total and LDL cholesterol, which are always related in real patients.
- **Inconsistent relationships.** Memory complaints are linked to diagnosis, and diagnosis is linked to lower MMSE, but memory complaints are not linked to MMSE. Some known risk factors (family history, diabetes) even go slightly in the opposite direction.
- **Coding uncertainties.** Gender, Ethnicity and EducationLevel are stored as numbers without labels in the file. I used the labels from the dataset description, but cannot confirm them from the data itself.
- **Missing context.** There is no information on how patients were recruited, how diagnosis was made, or when each variable was measured.

Taken together, these points suggest the dataset is synthetic. It is fine for practising analysis methods, but any patterns found in it should not be read as evidence about real Alzheimer's patients.
